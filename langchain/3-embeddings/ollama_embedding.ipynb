```python
# OllamaEmbeddings is a class from langchain_community that provides an interface to generate vector embeddings for text using Ollama models. 
# These embeddings are numerical representations of text, useful for semantic search, similarity comparison, and other NLP tasks.

# Real-time use cases:
# 1. Semantic Search: Quickly find relevant documents or passages by comparing embeddings of queries and documents.
# 2. Question Answering: Retrieve contextually similar answers from a knowledge base using embedding similarity.
# 3. Recommendation Systems: Suggest similar items, articles, or products based on embedding proximity.
# 4. Clustering & Classification: Group or categorize text data in real time by analyzing their embeddings.
# 5. Chatbots & Virtual Assistants: Enhance understanding of user queries and context for better responses.

from langchain_community.embeddings import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="gemma:2b")
embeddings

```

```python
r1 = embeddings.embed_query("What is the capital of France?")
r1
```

```python
r1 = embeddings.embed_documents(
    ["Alpha is the first letter of Greek alphabet", 
    "Beta is the second letter of Greek alphabet",
    "Gamma is the third letter of Greek alphabet"]
    )
r1[1]    
```

```python
r2 = embeddings.embed_query("What is the second letter of Greek alphabet?")
r2
```

```python
# mxbau-embed-large model
embeddings2 = OllamaEmbeddings(model="mxbai-embed-large")
embeddings2
text = "The capital of France is Paris."
r3 = embeddings2.embed_query(text)
r3
```
