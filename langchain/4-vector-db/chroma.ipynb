```python
# Summary and Real-time Use Cases for ChromaDB

"""
ChromaDB is an open-source embedding database designed for building applications that require efficient storage, retrieval, and search of vector embeddings. It is commonly used in AI, machine learning, and natural language processing workflows, especially for tasks involving semantic search, recommendation systems, and retrieval-augmented generation (RAG).

Summary:
- ChromaDB stores and manages high-dimensional vector embeddings.
- It enables fast similarity search and retrieval of relevant data points.
- Integrates easily with LLMs and other AI models for enhanced search and reasoning.

Real-time Use Cases:
1. Semantic Search: Powering search engines that return results based on meaning rather than keyword matching.
2. Recommendation Systems: Suggesting products, articles, or content based on user preferences and behavior.
3. Chatbots & Virtual Assistants: Retrieving relevant context or documents to provide accurate responses.
4. Document Deduplication: Identifying and removing duplicate or near-duplicate documents in large datasets.
5. Fraud Detection: Matching transaction patterns or user behaviors to detect anomalies in real-time.
6. Personalized Learning: Recommending educational resources tailored to a learner’s needs and progress.
7. Retrieval-Augmented Generation (RAG): Supplying LLMs with relevant context from a knowledge base to improve answer quality.

ChromaDB is especially useful in scenarios where low-latency, high-throughput vector search is required.
"""
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.document_loaders import TextLoader
from langchain.embeddings import OllamaEmbeddings
from langchain_chroma import Chroma
```

```python
loader = TextLoader("speech.txt")
documents = loader.load()
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
texts = text_splitter.split_documents(documents)
embeddings = OllamaEmbeddings(model="llama2")
db = Chroma.from_documents(texts, embeddings, persist_directory="chroma_db")
```

```python
query = "What is the main topic of the speech?"
docs = db.similarity_search(query)
docs[0].page_content

```

```python
#Saving to disk
vector_store = Chroma.from_documents(texts, embeddings, persist_directory="chroma_db")
```

```python
# Load from disk
vector_store2 = Chroma(persist_directory="chroma_db", embedding_function=embeddings)
```

```python
#Search
docs = vector_store2.similarity_search(query)
docs[0].page_content
```

```python
# Search with scores
docs_with_scores = vector_store2.similarity_search_with_score(query)
docs_with_scores
```

```python
# Retriver
retriever = vector_store2.as_retriever(search_type="similarity", search_kwargs={"k": 3})
docs = retriever.invoke(query)
docs[0].page_content
```
