```python
# HTMLHeaderTextSplitter is a utility that splits HTML documents into sections based on header tags (e.g., <h1>, <h2>, etc.).
# This is useful for processing, analyzing, or indexing content by logical sections.

# Real World Examples:
# 1. Extracting chapters or sections from online documentation or articles for summarization.
# 2. Preprocessing HTML content for search engines to improve indexing by section.
# 3. Splitting blog posts or news articles into sections for display in a table of contents or navigation sidebar.
```

```python
from langchain.text_splitter import HTMLHeaderTextSplitter

html_content = """
<html>
<head>
<title>Sample Document</title>
</head>
<body>
<h1>Introduction</h1>
<p>This is the introduction section of the document.</p>
<h2>Background</h2>
<p>This section provides background information.</p>
<h2>Methods</h2>
<p>Details about the methods used in the study.</p>
<h1>Results</h1>
<p>Here are the results of the study.</p>
<h2>Analysis</h2>
<p>Analysis of the results.</p>
<h1>Conclusion</h1>
<p>This is the conclusion of the document.</p> 
</body>
</html>
"""
html_headers = [("h1", "Header 1"), ("h2", "Header 2"), ("h3", "Header 3")]

  # Specify which header tags to split on
html_splitter = HTMLHeaderTextSplitter(html_headers)
html_chunks = html_splitter.split_text(html_content)
html_chunks
```

```python
# With url
url = "https://plato.stanford.edu/entries/goedel/"

header_tags = [("h1", "Header 1"), ("h2", "Header 2"), ("h3", "Header 3")]
html_splitter_with_url = HTMLHeaderTextSplitter(header_tags)
html_chunks_with_url = html_splitter_with_url.split_text(url)
html_chunks_with_url
```
