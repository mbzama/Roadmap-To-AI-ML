```python
## Models

from pydantic import BaseModel

class Person(BaseModel):
    name: str
    age: int
    city: str


```

```python
## Data validation with Pydantic
# valid
person = Person(name="zama", age="60", city="Hyderabad")
print(person)


# invalid
person = Person(name="zama", age="60", city=555555)
print(person)
```

```python
## Without pydantic

class PersonWithoutPydantic:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city

person2 = PersonWithoutPydantic(name="zama", age="60", city="Hyderabad")
print(person2.__dict__)

person2 = PersonWithoutPydantic(name="zama", age="60", city=55555)
print(person2.__dict__)

```

```python
## Model with optional fileds

from typing import Optional

class Employee(BaseModel):
    name: str
    age: int
    department: str
    salary: Optional[float] = 0.0
    is_active: Optional[bool] = True

emp1 = Employee(name="John", age=30, department="Engineering")
print(emp1)    
```

```python
## list
from pydantic import BaseModel
from typing import List

class Classroom(BaseModel):
    name: str
    students: List[str]
    teacher: str
    capacity: int
```

```python
# Valid
classroom = Classroom(
    name="Math 101",
    students=["Alice", "Bob", "Charlie"],
    teacher="Mr. Smith",
    capacity=30
)
print(classroom)
```

```python
# Invalid
classroom = Classroom(
    name="Math 101",
    students=["Alice", "Bob", 123],  # Invalid student name
    teacher="Mr. Smith",
    capacity=30 # Invalid capacity type
)

try:
    print(classroom)
except Exception as e:
    print(f"Error: {e}")


```

```python
## nested models
from pydantic import BaseModel

class Address(BaseModel):
    street: str
    city: str
    state: str
    zip_code: int

class User(BaseModel):
    id: int
    name: str
    email: str
    address: Address

#Valid
user = User(
    id=1,
    name="John Doe",
    email="john@gmail.com",
    address=Address(
        street="123 Main St",
        city="Anytown",
        state="CA",
        zip_code=12345
    )
)
print(user)   
```

```python
#Valid
user = User(
    id=1,
    name="John Doe",
    email="john@gmail.com",
    address=Address(
        street="123 Main St",
        city="Anytown",
        state="CA",
        zip_code="12345"
    )
)
print(user)   
```

```python
# custom validation
from pydantic import BaseModel, Field

class Item(BaseModel):
    name: str
    price: float = Field(gt=0, description="Price must be greater than 0")
    quantity: int = Field(ge=0, le=100, description="Quantity must be 0 or greater")

# Valid
item = Item(name="Apple", price=1.5, quantity=10)
print(item)
```

```python
#Imvalid
try:
    item = Item(name="Banana", price=-1.0, quantity=150)  # Invalid price
except ValueError as e:
    print(f"Error: {e}")
 
```

```python
# schema definition
print(Item.schema_json(indent=2))
```
